```python
import sys
sys.path.append("..")
from pathlib import Path
import pandas as pd
from src.features import build_features

load = pd.read_parquet("../data/raw/kse_load.parquet")
weather = pd.read_parquet("../data/raw/weather_pl.parquet")

features = build_features(load, weather)
Path("../data/processed").mkdir(parents=True, exist_ok=True)
features.to_parquet("../data/processed/features.parquet")
print(features.shape)
features.head()
```

```python
def mae(pred):
    return (features["load_actual"] - pred).abs().mean().round(0)

print("MAE prognozy PSE:          ", mae(features["load_fcst"]), "MW")
print("MAE 'jak tydzień temu':    ", mae(features["load_lag_168h"]), "MW")
print("MAE 'jak 2 dni temu':      ", mae(features["load_lag_48h"]), "MW")
```
