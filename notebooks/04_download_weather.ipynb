```python
import sys
sys.path.append("..")
from pathlib import Path
import pandas as pd
from src.weather_client import fetch_weather, weighted_average

wczoraj = (pd.Timestamp.today() - pd.Timedelta(days=1)).strftime("%Y-%m-%d")
raw = fetch_weather("2024-06-14", wczoraj)

Path("../data/raw").mkdir(parents=True, exist_ok=True)
raw.to_parquet("../data/raw/weather_cities.parquet", index=False)

weather = weighted_average(raw)
weather.to_parquet("../data/raw/weather_pl.parquet")
print(raw.shape, weather.shape)
weather.tail()
```

```python
print("Zakres:", weather.index.min(), "->", weather.index.max())
pelny = pd.date_range(weather.index.min(), weather.index.max(), freq="h")
print("Brakujące godziny:", len(pelny.difference(weather.index)))
print("Puste wartości w surowych danych:")
print(raw.isna().sum())
```
